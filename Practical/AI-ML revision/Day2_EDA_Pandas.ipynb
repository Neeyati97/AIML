45. Detect Outliers using IQR

```python
Q1 = df["Annual_salary"].quantile(0.25)
Q3 = df["Annual_salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Annual_salary"] < lower) |
    (df["Annual_salary"] > upper)
]
outliers
```

46. Filter without outliers

```python
df_clean = df[
    (df["Annual_salary"] >= lower) &
    (df["Annual_salary"] <= upper)
]
```
